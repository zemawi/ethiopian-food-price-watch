# Ethiopia Food Prices — 11 MODEL COMPARISON

Compare forecasting and spike detection models using the metric that matches each business objective.
For forecasting, we minimize error; for spike detection, we maximize F1 while keeping ROC-AUC strong.


```python
import pandas as pd
from pathlib import Path

root = Path('../')
forecast_df = pd.read_csv(root / 'models' / 'forecast_metrics.csv')
spike_df = pd.read_csv(root / 'models' / 'spike_metrics.csv')

print('Forecasting comparison (lower RMSE is better)')
print(forecast_df.sort_values('rmse').to_string(index=False))
print('\nSpike classification comparison (higher F1 is better)')
print(spike_df.sort_values('f1', ascending=False).to_string(index=False))
```

```python
best_forecast = forecast_df.sort_values('rmse').iloc[0].to_dict()
best_spike = spike_df.sort_values('f1', ascending=False).iloc[0].to_dict()

print('Production recommendation')
print(f'Best forecasting model: {best_forecast["model"]} with RMSE {best_forecast["rmse"]:.6f}')
print(f'Best spike model: {best_spike["model"]} with F1 {best_spike["f1"]:.4f}, ROC-AUC {best_spike["roc_auc"]:.4f}, threshold {best_spike["best_threshold"]:.2f}')
print('\nWhy this is the correct decision:')
print('1. Forecasting is a regression problem, so RMSE is the right business metric.')
print('2. Spike detection is a class-imbalance problem, so F1 and ROC-AUC are more meaningful than accuracy.')
print('3. The selected decision threshold is tuned on validation data, which is more production-safe than using the default 0.5 cutoff.')
```
